```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
```

```python
from vnpy_ctastrategy.backtesting import BacktestingEngine

from datetime import datetime
import pickle
# from strategies.pxc_test_strategy import SilverT
from strategies.silver_t_backtest import SilverT

capital = 3000000

# 重新创建回测引擎
enging = BacktestingEngine()

enging.set_parameters(
    vt_symbol="ag88.SHFE",
    interval="1m",
    start=datetime(2025, 1, 1),
    end=datetime(2025, 1, 2),
    rate=0.3 / 10000,
    slippage=1,
    size=10,
    pricetick=1,
    capital=capital,
)

# 加载数据
with open("../../data/白银主连数据/2025年/分钟级/ag88_1m_250101_251027.pkl", "rb") as f:
    enging.history_data = pickle.load(f)

enging.add_strategy(SilverT, {"account_equity":capital})

# 运行回测
enging.run_backtesting()

import logging
import time

logs = enging.logs
# write logs to a file
logging.basicConfig(filename=f"backtesting_logs_{int(time.time())}.log", level=logging.DEBUG, format='%(levelname)s - %(message)s')
for log in logs:
    logging.debug(log)
```

```python
# 显示回测结果
import pandas as pd
enging.output = lambda x: print(x)  # 不格式化输出
# Save results to a CSV file
enging.calculate_result()
enging.calculate_statistics()
results_df = pd.DataFrame(enging.daily_df)
results_df.to_csv("backtesting_results.csv")

```
